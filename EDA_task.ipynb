```python

# Step 6: Outlier Detection

# 1. Check outliers in Profit
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Profit"])
plt.title("Outlier Detection in Profit")
plt.xlabel("Profit")
plt.show()

# 2. Check outliers in Revenue
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Revenue"])
plt.title("Outlier Detection in Revenue")
plt.xlabel("Revenue")
plt.show()

# 3. Check outliers in Ad_Spend
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Ad_Spend"])
plt.title("Outlier Detection in Ad Spend")
plt.xlabel("Ad Spend")
plt.show()

# 4. Check outliers in Units_Sold
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Units_Sold"])
plt.title("Outlier Detection in Units Sold")
plt.xlabel("Units Sold")
plt.show()
```

```python

# Step 6: Verify Outliers Using IQR Method

for col in ["Profit", "Revenue"]:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\n{col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
    print(outliers[[col]].sort_values(by=col))
```

### Outlier Detection

The IQR method was used to identify potential outliers in the dataset.

- **Profit:** 38 potential outliers were identified.
- **Revenue:** 6 potential outliers were identified.
- **Ad_Spend:** No visible outliers were observed in the boxplot.
- **Units_Sold:** No visible outliers were observed in the boxplot.

The potential outliers were retained because they may represent genuine business performance rather than data errors. Further investigation would be needed before removing or modifying any values.

```python

# Step 7: Feature Engineering

# 1. Create Profit Margin (%)
df["Profit_Margin"] = np.where(
    df["Revenue"] != 0,
    (df["Profit"] / df["Revenue"]) * 100,
    np.nan
)

# 2. Create Advertising Spend as a Percentage of Revenue
df["Ad_Spend_Percentage"] = np.where(
    df["Revenue"] != 0,
    (df["Ad_Spend"] / df["Revenue"]) * 100,
    np.nan
)

# 3. Display the new features
print("Dataset with New Features:")
print(
    df[
        [
            "Revenue",
            "Profit",
            "Profit_Margin",
            "Ad_Spend",
            "Ad_Spend_Percentage"
        ]
    ].head()
)

# 4. Check the updated dataset shape
print("\nUpdated Dataset Shape:", df.shape)
```

```python

# Business Question 3: Profit Margin by Category

category_margin = (
    df.groupby("Category")
    .agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Profit=("Profit", "sum")
    )
)

category_margin["Profit_Margin_Percent"] = (
    category_margin["Total_Profit"] /
    category_margin["Total_Revenue"]
) * 100

category_margin = category_margin.sort_values(
    "Profit_Margin_Percent",
    ascending=False
)

print("Profit Margin by Category:")
print(category_margin.round(2))

plt.figure(figsize=(9, 5))
sns.barplot(
    data=category_margin.reset_index(),
    x="Category",
    y="Profit_Margin_Percent"
)

plt.title("Profit Margin by Product Category")
plt.xlabel("Category")
plt.ylabel("Profit Margin (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Business Question 3: Which Category Has the Highest Profit Margin?

**Observation:**
- Groceries has the highest profit margin at **21.14%**.
- Beauty Products has the lowest profit margin at **19.56%**.
- The profit margins of all six categories are relatively close, ranging from 19.56% to 21.14%.

**Insight:**
Groceries generates the highest profit margin, meaning it retains a larger proportion of its revenue as profit compared with the other categories. However, Home Decor generates the highest total revenue, while Groceries generates the highest total profit.

**Business Decision:**
The business should investigate what makes Groceries more profitable and consider applying successful strategies to other categories. It should also monitor costs, sales volume, and customer demand before making investment decisions.
